```python
import pandas as pd
import os

# 设置你的根目录路径 (根据 image_bb781d.png 中的路径)
root_path = r'C:\python\PYTHON\project_folder\cleaned_data'

# 创建一个字典来存储所有的 DataFrame
# 键 (Key) 是文件名，值 (Value) 是对应的 DataFrame 数据
all_data = {}

for root, dirs, files in os.walk(root_path):
    for file in files:
        if file.endswith('.csv'):
            # 构建完整文件路径
            file_path = os.path.join(root, file)
            # 读取并存入字典，去掉 .csv 后缀作为名称
            table_name = file.replace('.csv', '')
            all_data[table_name] = pd.read_csv(file_path)
            print(f"已加载表: {table_name}")

# 现在你可以直接通过名字访问特定的数据
# 比如查看第一个文件
# print(all_data['annual-value-and-property-tax-by-property-type-annual'].head())
```

```python
import matplotlib.pyplot as plt

import seaborn as sns

# 目标表格列表
target_tables = [
    'median-annual-value-and-property-tax-by-type-of-hdb',
    'median-annual-value-and-property-tax-by-type-of-private-residential-property'
]

def analyze_growth_drivers(table_name):
    if table_name not in all_data:
        print(f"找不到表: {table_name}")
        return

    df = all_data[table_name].copy()

    # 数据转换与清洗
    df['financial_year'] = pd.to_numeric(df['financial_year'], errors='coerce')
    df['property_tax_collection'] = pd.to_numeric(df['property_tax_collection'], errors='coerce')
    df['no_of_cases'] = pd.to_numeric(df['no_of_cases'], errors='coerce')

    # 聚合年度数据
    annual_df = df.groupby('financial_year').agg({
        'property_tax_collection': 'sum',
        'no_of_cases': 'sum'
    }).reset_index()

    # 计算关键指标：单位税收 (反映单套房产价值/税率变化)
    annual_df['tax_per_case'] = annual_df['property_tax_collection'] / annual_df['no_of_cases']

    # 计算增长率 (YoY)
    annual_df['tax_growth'] = annual_df['property_tax_collection'].pct_change()
    annual_df['cases_growth'] = annual_df['no_of_cases'].pct_change()

    # 可视化
    fig, ax1 = plt.subplots(figsize=(12, 6))

    # 1. 绘制总税收增长曲线
    sns.lineplot(data=annual_df, x='financial_year', y='property_tax_collection',
                 ax=ax1, marker='o', label='Total Property Tax', color='blue', linewidth=2)
    ax1.set_ylabel('Total Property Tax ($)', color='blue')

    # 2. 绘制单位税收趋势 (Secondary Axis)
    ax2 = ax1.twinx()
    sns.lineplot(data=annual_df, x='financial_year', y='tax_per_case',
                 ax=ax2, marker='s', label='Tax per Case (AV Proxy)', color='red', linestyle='--')
    ax2.set_ylabel('Tax per Case ($)', color='red')

    plt.title(f'Growth Drivers Analysis: {table_name}')
    ax1.grid(True, alpha=0.3)
    plt.show()

    # 打印相关性分析
    corr = annual_df[['property_tax_collection', 'no_of_cases', 'tax_per_case']].corr()['property_tax_collection']
    print(f"\n--- {table_name} 驱动力诊断 ---")
    print(f"税收与【房产数量】的相关性: {corr['no_of_cases']:.4f}")
    print(f"税收与【单套价值指标】的相关性: {corr['tax_per_case']:.4f}")

# 执行分析
for table in target_tables:
    analyze_growth_drivers(table)
```

```python
fiscal_files = [
    'government-fiscal-position',
    'government-operating-revenue',
    'government-total-expenditure',
    'government-manpower',
    'government-expenditure-by-type'
]

# 预处理函数：统一列名和类型
def clean_fiscal_data(name):
    df = all_data[name].copy()
    df.columns = df.columns.str.strip()
    # 查找包含年份的列并统一命名
    year_col = [c for c in df.columns if 'year' in c or 'date' in c][0]
    df.rename(columns={year_col: 'year'}, inplace=True)
    df['year'] = pd.to_numeric(df['year'], errors='coerce')
    if 'amount' in df.columns:
        df['amount'] = pd.to_numeric(df['amount'], errors='coerce')
    return df

# --- 维度 1: 财政健康度 (Revenue vs Expenditure) ---
fiscal = clean_fiscal_data('government-fiscal-position')
revenue = fiscal[fiscal['category'] == 'Revenue'].groupby('year')['amount'].sum()
expense = fiscal[fiscal['category'] == 'Expenditure'].groupby('year')['amount'].sum()

plt.figure(figsize=(12, 6))
plt.plot(revenue.index, revenue.values, label='Total Revenue', marker='o', linewidth=2)
plt.plot(expense.index, expense.values, label='Total Expenditure', marker='s', linewidth=2)
plt.fill_between(revenue.index, revenue.values, expense.values, color='gray', alpha=0.1, label='Fiscal Gap/Surplus')
plt.title('Singapore Fiscal Health: Revenue vs Expenditure')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

# --- 维度 2: 人力资源扩张 (Government Manpower) ---
manpower = clean_fiscal_data('government-manpower')
top_ministries = manpower.groupby('ministry')['number'].mean().sort_values(ascending=False).head(5)
print("\n--- 平均雇员最多的 5 个部门 ---")
print(top_ministries)

# --- 维度 3: 支出结构重心 (Expenditure Sector) ---
exp_total = clean_fiscal_data('government-total-expenditure')
sector_trend = exp_total.groupby(['year', 'sector'])['amount'].sum().unstack()
sector_trend.plot(kind='area', stacked=True, figsize=(12, 6), alpha=0.7)
plt.title('Expenditure Evolution by Sector')
plt.ylabel('Amount ($)')
plt.show()
```

```python
import pandas as pd
import os
import matplotlib.pyplot as plt
import seaborn as sns

# --- 1. 路径自动定位 ---
# 获取当前脚本所在目录
base_dir = os.path.dirname(os.path.abspath(__name__))

# 设定相对路径（适配你的文件夹结构）
data_folder = os.path.join(base_dir, "project_folder", "cleaned_data")

# 定义具体文件路径
path_base = os.path.join(data_folder, "yearly-interest-credited-to-cpf-members",
                         "yearly-amount-of-base-interest-credited-to-cpf-members.csv")
path_extra = os.path.join(data_folder, "yearly-interest-credited-to-cpf-members",
                          "yearly-amount-of-extra-interest-credited-to-cpf-members.csv")

def load_and_plot_cpf():
    try:
        # --- 2. 加载数据 ---
        df_base = pd.read_csv(path_base)
        df_extra = pd.read_csv(path_extra)

        # 统一列名（假设原始列名为 year 和 amount，根据实际情况调整）
        # 这里使用索引获取，确保即使列名微差也能运行
        df_base.columns = ['year', 'base_interest']
        df_extra.columns = ['year', 'extra_interest']

        # 合并数据并计算总额
        df_total = pd.merge(df_base, df_extra, on='year')
        df_total['total_interest'] = df_total['base_interest'] + df_total['extra_interest']
        df_total['extra_pct'] = (df_total['extra_interest'] / df_total['total_interest']) * 100

        print("✅ 数据加载成功！开始生成图表...")

        # --- 3. 可视化逻辑 ---
        sns.set_theme(style="white")
        plt.rcParams['font.sans-serif'] = ['Arial Unicode MS', 'SimHei', 'sans-serif']
        plt.rcParams['axes.unicode_minus'] = False

        fig, ax1 = plt.subplots(figsize=(12, 7))

        # 绘制堆叠柱状图 (显示利息总额)
        ax1.bar(df_total['year'], df_total['base_interest'], label='Base Interest (基础利息)', color='#5DADE2')
        ax1.bar(df_total['year'], df_total['extra_interest'], bottom=df_total['base_interest'],
                label='Extra Interest (额外利息)', color='#EB984E')

        ax1.set_xlabel('Financial Year (财政年度)', fontsize=12)
        ax1.set_ylabel('Interest Amount (利息总额)', fontsize=12)
        ax1.legend(loc='upper left')

        # 绘制折线图 (显示额外利息的占比趋势)
        ax2 = ax1.twinx()
        sns.lineplot(data=df_total, x='year', y='extra_pct', ax=ax2, color='red', marker='d', label='Extra %')
        ax2.set_ylabel('Extra Interest % (额外利息占比)', color='red')
        ax2.set_ylim(0, max(df_total['extra_pct']) * 1.5)
        ax2.grid(False)

        plt.title('Singapore CPF Interest Distribution Trend\n(Annual Base vs Extra Interest)', fontsize=14, pad=15)
        plt.tight_layout()

        # 保存图片
        plt.savefig('cpf_interest_analysis.png', dpi=300)
        plt.show()

        print("📈 分析报告已生成：'cpf_interest_analysis.png'")

    except FileNotFoundError:
        print("❌ 找不到文件！")
        print(f"请检查路径是否正确: {os.path.abspath(data_folder)}")
    except Exception as e:
        print(f"❌ 运行出错: {e}")

if __name__ == "__main__":
    load_and_plot_cpf()
```
